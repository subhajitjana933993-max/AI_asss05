```python
from collections import deque

goal = "123456780"

def bfs(start):
    q = deque([(start, 0)])
    visited = {start}

    while q:
        s, level = q.popleft()

        print("Level", level, ":", s)

        if s == goal:
            print("Solved at Level", level)
            return

        i = s.index("0")

        moves = [
            (1, 3),
            (0, 2, 4),
            (1, 5),
            (0, 4, 6),
            (1, 3, 5, 7),
            (2, 4, 8),
            (3, 7),
            (4, 6, 8),
            (5, 7)
        ]

        for j in moves[i]:
            t = list(s)
            t[i], t[j] = t[j], t[i]
            t = "".join(t)

            if t not in visited:
                visited.add(t)
                q.append((t, level + 1))

    print("Not Solved")


start = input("Enter puzzle: ")
bfs(start)
```
